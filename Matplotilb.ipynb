```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()         #display graph
```

```python
import matplotlib.pyplot as plt
month = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.plot(month,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
month = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
profit = [5,8,6,12,15]
plt.plot(month,sales,label="Sales")   #here the label means name the label
plt.plot(month,profit,label="Profit")
plt.legend()   #it means display the names
plt.show()
```

```python
import matplotlib.pyplot as plt
month = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.plot(month,sales)
plt.xlabel("month")
plt.ylabel("sales")
plt.grid(True)   #grid varan
plt.plot(
    month,
    sales,
    marker="s",      #show individual data point
    color="red",
    linestyle="--"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
month = ["Jan","Feb","Mar","Apr","May"]
sales = [20,30,25,40,50]
plt.bar(month,sales)
plt.xlabel("month")
plt.ylabel("sales")
plt.grid(True)   #grid varan
plt.plot(
    month,
    sales,
    marker="s",      #show individual data point
    color="red",
    linestyle="--"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
study_hours = [1,2,3,4,5,6]
marks = [40,45,55,65,75,85]
plt.scatter(study_hours,marks)
plt.title("Study Hours Vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
marks = [40,45,55,65,75,100]
plt.hist(marks)
plt.hist(study_hours,marks)
plt.title("Study Hours Vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
#pie chart
categories = ["Food","Trave","Shopping","Others"]
amount = [4000,3000,2000,1000]
plt.pie(amount, labels=categories)
plt.title("Monthly Expenses")
plt.legend()
plt.show()
```

```python
# ax means individual graph
import matplotlib.pyplot as plt
fig, ax = plt.subplots(2,1)
ax[0].plot([1,2,3],[10,20,30])
ax[1].bar(["A","B","C"],[20,15,30])
plt.show()
```
