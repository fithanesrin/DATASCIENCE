question 1 - online store sales

an online store has recorded the no of units sold for different products.

the manager wants to compare the performance of the products and easily identify which product sold the most.

task:visualize the data in a suitable way



```python
import matplotlib.pyplot as plt
products = ["laptop","phone","tablet","watch"]
sales = [50,80,40,70]
plt.bar(products,sales)
plt.show()
```

2. website visitors

company recorded its website visitors during a working week

the marketing team wants to understand how visistor activity changed throughout the week

task:create a suitable visualization to show this change

```python
days = ["mon","tue","wed","thu","fri"]
visitors = [120,150,180,160,210]
plt.plot(days,visitors,marker="*")
plt.xlabel("days")
plt.ylabel("visitors")
plt.title("website visitors")
plt.show()
plt.fill_between(days,visitors)
plt.show()

```

3. monthly expenses

a student recorded their monthly expenses

the student wants to understand how the total spending is dividing among the different categories

task:represent the information visually

```python
categories = ["Food","Traver","Rent","Shopping"]
expenses = [5000,3000,8000,4000]
plt.pie(expenses,labels=categories,autopct="%1.1f%%")
plt.show()
```

4. advertising and sales

an e-commerce company recorded its advertising expenditure and sales.

the company wants to investigate whether changes in advertising expenditure are associate with changes in sales

task: create a suitable visualization for this analysis

```python
advertising = [10,20,30,40,50]
sales = [25,35,45,55,70]
plt.scatter(advertising,sales)
plt.xlabel("advertising")
plt.ylabel("sales")
plt.show()
```
