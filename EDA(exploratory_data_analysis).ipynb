```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/content/titanic.csv')
df
```

```python
df.head()
```

```python
print(df.shape)  #to show no of rows and columns
```

```python
df.info()   #to show columns name,data type
```

```python
df.describe()   # roe and column give mean ,median,mode ,average(statistical  summurary)
```

```python
df.isnull().sum()   # find null values ,sum- numbers of errors
```

```python
df["Embarked"] = df["Embarked"].fillna(    #replace missing value
   df["Embarked"].mode()[0]    #find the most frequent value
)
print(df.isnull().sum())
print("before:",df.shape[0])
df=df.drop_dulicates()
print("after:",df.shape[0])
```

```python
df = df.rename(columns={
    "Passengerid":"PassengerId",
    "2urvived":"Survived"
})
df
```

```python
df["Sex"] = df["Sex"].map({
    0:"Male",
    1:"female"
})
df
```

```python
df.head()
sns.histplot(data=df,x="Age",bins=20)
```

```python
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
sns.countplot(data=df,x="Sex",hue="Survived")

plt.title("Titanic survival by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.show()
sns.countplot(data=df,x="Pclass",hue="Survived")

plt.title("Titanic survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```
