```python
import pandas as pd
data=[10,20,30,40,50]       #Series
s=pd.Series(data)
print(s)
```

```python
import pandas as pd
data=[10,20,30,40,50]       #index
s=pd.Series(data,index=['A','B','C','D','E'])
print(s)
```

loc and iloc  is used to access and slicing

iloc=INTEGER

loc=LABEL BASED

```python
data=[60,70,80,90,]
s=pd.Series(data,index=['a','b','c','d'])
print(s.loc['a'])
print(s.iloc[1])
```

```python
# slicing
data=[10,20,30,40,50]
s=pd.Series(data,index=['a','b','c','d','e'])
print(s.loc['a':'b'])
print(s.iloc[:2])
```

```python
import pandas as pd
sales=pd.Series([1200,850,1500,950,2000],
                index=["monday","tuesday","wednesday","thursday","friday"])
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
print(sales[sales>1000])
```

```python
#unique
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())
```

```python
#count
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.nunique())
print(ar.value_counts())  # repeated value count
```

```python
data=[20,None,30,30,None,40,50,60]
ar=pd.Series(data)
print(ar)
print(ar.isnull())
print(ar.notnull())
```

```python
#none value removing
data=[20,None,30,30,None,40,50,60]
ar=pd.Series(data)
print(ar)
print(ar.isnull())
print(ar.notnull())
print(ar.dropna())
```
