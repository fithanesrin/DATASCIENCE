NUMPY(Numeric python):
numpy is a python library used for working with arrays
import:import numpy as np

used for mathematical operations

```python
a=[1,2,3,4,5]
b=[6,7,8,9,10]
add=(a+b)
print(add)
```

```python
import numpy as np
arr1=np.array([1,2,3,4,5])       # mathematial use anu numpy
arr2=np.array([6,7,8,9,10])
result=(arr1+arr2)
print(result)
```

```python
# indexing
array=[10,20,30,40,50]
print(array[3])
```

```python
# slicing
# one dimension
arr1=np.array([10,20,30,40,50,60,70,80])
print(arr1.ndim)
print(arr1[1:8:2])
```

```python
# 2 dimensinal array
arr1=np.array([[10,20,30,40],[50,60,70,80]])
print(arr1.ndim)


```

```python
# 3 dimension
arr1=np.array([[[10,20,30],[40,50,60],[70,80,90]]])
print(arr1.ndim)
```

```python
arr1=np.array([[[10,20,30],
                [40,50,60]],

                [[70,80,90],
                [25,35,45]]])
print(arr1[0,1,2])
```

```python
# one dimension convert to 2dimension/reshape
arr1=np.array([1,2,3,4,5,6,7,8])
arr2=arr1.reshape(2,4)
print(arr1)
print(arr2)
print(arr2.shape)
```

```python
# function-zeros
a=np.zeros(5)
print(a)
```

```python
# function ones
a=np.ones(5)
print(a)
```

```python
# function arrange
c=np.arange(1,10)
print(c)
```

```python
#linspace
d=np.linspace(0,20,4)
print(d)
```

```python
# random
e=np.random.rand(5)
print(e)
```

SERIES:


*Series is an one dimension labelled array

```python
import pandas as pd
mark=pd.Series([1,2,3,4,5,6])
print(mark)
```

```python
import pandas as pd
marks=pd.Series([50,79,80,90,],index=["maths","physics","biology","chemistry"])
print(marks)
```

```python
import pandas as pd
name=pd.Series(["language","computer","commerce","science"],index=["fitha","shifa","shanidha","ardra"])
print(name)
```

```python
import pandas as pd
name=pd.Series(["language","computer","commerce","science"],index=["fitha","shifa","shanidha","ardra"])
print(name["shanidha"])
```
