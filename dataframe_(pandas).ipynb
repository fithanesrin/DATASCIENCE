```python
#create a dataframe
import pandas as pd
import numpy as np
data = {
    "name":["alice","bob","charlie","david","emma"],
    "age":[20,21,19,22,20],
    "marks":[85,72,np.nan,90,65],
    "attendance":[95,80,75,np.nan,88],
    "city":["kochi","delhi","mumbai","kochi","chennai"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# first 3 rows print cheyan df.head()
#last rows print cheyan df.tail()
print("\n fist 3 rows:")
print(df.head(3))


print("\n last 2 rows:")
print(df.tail(2))
```

```python
#shape- datas arrange cheyan
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df["marks"])    # only marks
print(df[["name" ,"marks"]])   #marks and name
print("\t\t\t----------------------------------------------------\t\t\t")
```

```python
#slicing
print("\n first 3 roes:")
print(df.iloc[0:3])        #integer base,complete kittulaaa
print("\n rows 1 to 3, name and marks:")
print(df.loc[1:3,["name","marks"]])         #label or position based,complete kittum
```

```python
#filter
print("\n students with marks greater than 75:")
print(df[df["marks"]>75])

print("\n students from kochi:")
print(df[df["city"]=="kochi"])

print("\n students with marks >70 and attendance >80:")
print(df[(df["marks"]>70) & (df["attendance"]>80)])
```

```python
# mathematical operations
print("\nmarks + 5:")
print(df["marks"]+ 5)

print("\nmarks * 2:")
print(df["marks"]*2)

print("\naverage marks:")
print(df["marks"].mean())

print("\nmax marks:")
print(df["marks"].max())

print("\nmin marks:")
print(df["marks"].min())

print("\ntotal marks:")
print(df["marks"].sum())

```

```python
# sorting
print("\nsorted by marks:")
print(df.sort_values("marks"))

print("\nsorted by marks - highest first:")
print(df.sort_values("marks",ascending=False))
```

```python
# RENAMING COLUMNS
df_rename = df.rename(
    columns={
        "marks":"score",
        "attendance":"attendence_percentage"
    }
)
print("\nrenamed DataFrame:")
print(df_rename)
```

```python
df_renamed=df.rename(columns={"marks":"score","attendance":"attendence_percentage"})
print(df_renamed)
```

```python
#checking missing values
print("\nchecking missing values:")
print(df.isna())

print("\nNumber of missing values:")
print(df.isna().sum())

print(df.dropna())  # null remove
```

```python
#forward fill  - nulll ulla values remove akki mele ullath fill akum
df_forward = df.copy()
df_forward = df_forward.ffill()

print("\nforward fill:")
print(df_forward)

```

```python
#backward fill - null ulla value remove akki adinthe thazhe ullath no kond fill akum
df_backward = df.copy()
df_backward = df_backward.bfill()

print("\nbackward fill:")
print(df_backward)
```
